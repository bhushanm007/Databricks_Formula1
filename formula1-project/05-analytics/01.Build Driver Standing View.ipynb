## Build Driver Standings

#### Sources

1. fact_session_results
2. dim_drivers

#### Output Columns

1. season
2. driver_id
3. driver_name
4. nationality
5. race_starts
6. total_points
7. number_of_wins
8. number_of_podiums
9. standing_position

```sql
CREATE OR REPLACE VIEW formula1.gold.v_driver_standing
AS
WITH driver_session_summary
AS
    (SELECT 
            FS.season,
            dd.driver_id,
            dd.driver_name,
            dd.nationality,
            COUNT(*) AS race_start,
            SUM(FS.points) AS total_points,
            count_if(FS.is_win) AS number_of_wins,
            count_if(FS.is_podium) AS number_of_podiums 
    FROM formula1.gold.dim_drivers AS DD
    JOIN formula1.gold.fact_session_results AS FS
    ON DD.driver_id = FS.driver_id
    GROUP BY FS.season, dd.driver_id, dd.driver_name, dd.nationality)
SELECT RANK() OVER(PARTITION BY season ORDER BY total_points DESC, number_of_wins DESC) AS standing,
       season,
       driver_id,
       driver_name,
       nationality,
       race_start,
       total_points,
       number_of_wins,
       number_of_podiums
FROM driver_session_summary;
 
```

```sql
select *
from formula1.gold.v_driver_standing
where season = 2025;
```
