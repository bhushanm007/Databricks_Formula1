# Ingest constructors.json files
1. Read the files using spark dataframe reader API
2. Add Metadata Columns
    - Source File
    - Ingestion Timestamp
3. Write to bronze delta table

```python
source_file = f"{landing_folder_path}/constructors.json"
table_name = f"{catalog_name}.{bronze_schema}.constructors"
```

```python
constructors_schema = "constructorId string, name STRING, nationality STRING, url STRING"
```

```python
constuctors_df = (
    spark.read
        .format('json')
        .option('header', 'true')
        .option('mode', 'failFast')
        .schema(constructors_schema)
        .load(source_file)
)
```

cas

```python
display(constuctors_df)
```

```python
constructors_final_df = add_ingestion_metadata(constuctors_df)
```

```python
display(constructors_final_df)
```

```python
(  
    constructors_final_df
        .write
        .format('delta')
        .mode('overwrite')
        .option('overwriteSchema', 'true')
        .saveAsTable(table_name)
)
```
