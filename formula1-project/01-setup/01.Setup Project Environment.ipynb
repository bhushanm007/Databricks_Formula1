# Set-up the project environment for Formula1 Project
- 1. Create External Location databricks-project-ext-dl1-formula1
- 2. Create catalog formula1
- 3. Create Schemas landing, bronze, silver and gold
- 4. Create Volume Files in the landing schema 


## Access Cloud Storage

```sql
%fs ls 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/'
```

## Create External Location

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS databricks_project_ext_dl11_formula1
    URL 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL `databricks-project-sc`)
    COMMENT 'External location for the formula1 container';
```

```sql
%fs ls abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/landing/
```

## Create catalog formula1

```sql
SHOW CATALOGS;
```

```sql
CREATE CATALOG IF NOT EXISTS formula1
   MANAGED LOCATION 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/'
   COMMENT 'This is the main catalog for the formula1 project';
```

## Create Schemas landing, bronze, silver and gold

```sql
CREATE SCHEMA IF NOT EXISTS formula1.landing;
CREATE SCHEMA IF NOT EXISTS formula1.bronze
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS formula1.silver
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS formula1.gold
    MANAGED LOCATION 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/gold';        
```

```sql
USE CATALOG formula1
```

```sql
SELECT current_catalog();
```

```sql
show schemas;
```

## Create Volume Files in the landing schema

```sql
CREATE EXTERNAL VOLUME formula1.landing.files
LOCATION 'abfss://formula1@databricksprojectextdl11.dfs.core.windows.net/landing';
```

```sql
%fs ls /Volumes/formula1/landing/files
```
