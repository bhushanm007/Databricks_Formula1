### Build Races Dimension

1. Read silver races table
2. Read silver circuits table
3. Join the data from races with circuits using circuit_id
4. Select the required columns:
    * races.season
    * races.round
    * races.race_name
    * races.race_date
    * circuits.circuit_name
    * circuits.locality
    * circuits.country
5. Write the transformed data to gold dim_races table

```python
%run ../00-common/01.environment-configuration
```

```python
from pyspark.sql import functions as F
```

```python
target_table =F"{catalog_name}.{gold_schema}.dim_races"
```

Read source table
-  races
- circuits

```python
circuits_df = spark.table(f"{catalog_name}.{silver_schema}.circuits")
races_df = spark.table(f"{catalog_name}.{silver_schema}.races")
```

### Join the data from races with circuits using circuit_id
#### Select the required columns:
- races.season
- races.round
- races.race_name
- races.race_date
- circuits.circuit_name
- circuits.locality
- circuits.country

```python
dim_races_df =(
    races_df
    .join(circuits_df, races_df.circuit_id == circuits_df.circuit_id)
    .select(
        races_df.season,
        races_df.round,
        races_df.race_name,
        races_df.race_date,
        circuits_df.circuit_name,
        circuits_df.locality,
        circuits_df.country
    )   
)
```

```python
display(dim_races_df)
```

```python
(
    dim_races_df
    .write
    .format("delta")   
    .saveAsTable(target_table)
)
```
